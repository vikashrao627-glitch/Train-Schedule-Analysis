```python
# creating new column tenure_category

df_clean["tenure_category"] = pd.cut(
    df_clean["tenure"],
    bins=[-1, 12, 36, float("inf")],
    labels=["0-12 Months", "13-36 Months", "37+ Months"]
)
```

```python
tenure_counts = (
    df_clean["tenure_category"]
    .value_counts()
    .reindex(tenure_order)
)

tenure_counts
```

**Summary Table**

```python
tenure_summary = pd.DataFrame({
    "Customer_Count": tenure_counts
})

tenure_summary["Percentage"] = (
    tenure_summary["Customer_Count"]
    / tenure_summary["Customer_Count"].sum()
    * 100
)

tenure_summary
```

```python
print("Highest Churn Payment Method:", payment_churn.idxmax())
print("Churn Rate:", round(payment_churn.max(), 2), "%")
```

**Contract Finding**

```python
contract_churn = (
    df_clean.groupby("contract")["churn"]
    .apply(lambda x: (x == "Yes").mean() * 100)
    .sort_values(ascending=False)
)

print(contract_churn)
```

```python
print("Highest Churn Contract:", contract_churn.idxmax())
print("Churn Rate:", round(contract_churn.max(), 2), "%")
```
